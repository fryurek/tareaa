```python
"""
Misión: Crea un programa que simule la carga de una batería.
Instrucciones: Empieza con una variable carga = 0.
Mientras la carga sea menor a 100, suma de 20 en 20 e imprime el progreso.
Cuando llegue a 100, usa un if para mostrar "Carga Completa" y detén el proceso.
"""
carga = 0
print(f"Iniciando carga de batería...{carga}%")
while carga < 100:
    carga = carga + 20

    if carga == 100:
        print(f"Cargando... {carga}%")
        print("Carga Completa")

    else:
        print(f"Cargando...{carga}%")
```

```python
"""
Misión: Clasificación de temperatura de pacientes en tiempo real.
Instrucciones: El programa debe preguntar la temperatura de 5 pacientes (puedes
usar un contador para limitar los turnos).
Si la temperatura es >= 39, imprimir "Emergencia".Si está entre 37.5 y 38.9,
imprimir "Observación".
Si es menor, imprimir "Estable". Complejidad: El programa no debe cerrarse hasta
que se hayan evaluado los 5 pacientes.
"""
contador = 0
while contador < 5:
    contador = contador + 1
    temperatura = float(input(f"Ingrese la temperatura del paciente #{contador}: "))

    if temperatura >= 39:
        print("⚠️ Clasificación: Emergencia")

    elif 37.5 <= temperatura <= 38.9:
        print("👀 Clasificación: Observación")

    else:
        print("✅ Clasificación: Estable")
print("Monitoreo finalizado. Los 5 pacientes han sido evaluados.")
```
